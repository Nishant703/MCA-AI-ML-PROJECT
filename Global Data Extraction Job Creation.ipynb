```python
with open(configFile) as f:
    data = f.read()
configData = json.loads(data)
for i in range(len(configData.get("DB_Properties"))):
    config=configparser.ConfigParser()
    config.read(configData.get("DB_Properties_File"))
    srcConName=configData.get("DB_Properties")[i]
    print(config[srcConName]['host'])
    print(config[srcConName]['service_name'])
    connection=config[srcConName]['type']+"://"+config[srcConName]['user']+":"+config[srcConName]['password']+"@"+config[srcConName]['host']+"/"+ config[srcConName]['service_name']
    connection=connection.replace('"','')
    print(connection)
    filepath=configData.get("Output_File_Path")[i]+configData.get("Output_File_Name")[i]+configData.get("Output_File_Date_Setting")[i]+configData.get("Output_File_Extension")[i]
    print(filepath)
    engine = sqlalchemy.create_engine(connection)
    df_result = pd.read_sql(configData.get("QUERY")[i], engine)
    engine.dispose()
    print(type(df_result))
    if "xlsx" in configData.get("Output_File_Extension")[i]:
        df_result.to_excel(filepath, index=False)
    elif "xlsb" in configData.get("Output_File_Extension")[i]:
        print("NA")
        df_result.to_excel(filepath, index=False, engine='xlwings')
    elif "xls" in configData.get("Output_File_Extension")[i]:
        print("openpyxl")
        df_result.to_excel(filepath, index=False, engine='xlsxwriter')
        #df_result.to_excel(filepath, index=False, engine='openpyxl')
    else:
        df_result.to_csv(filepath, sep=configData.get("Output_File_Delimiter")[i], encoding="utf8",index=False)
    print("done")
    
```

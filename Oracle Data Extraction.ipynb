```python
import oracledb
import datetime
import configparser
import pandas as pd
from os import listdir
from os.path import isfile,join
from sqlalchemy import create_engine
import sqlalchemy
```

```python
configFile=r"C:/Users/nisha/OneDrive/Documents/Python_Scripts/Oracle_Data_Extraction/Config_File.ini"
srcConName="Oracle_UAT"
Spl_path=r"C:/Users/nisha/OneDrive/Documents/Python_Scripts/Oracle_Data_Extraction/SQLs"
```

```python
start_time=datetime.datetime.now()
print(start_time)
```

```python
config=configparser.ConfigParser()
config.read(configFile)
connection=config[srcConName]['type']+"://"+config[srcConName]['user']+":"+config[srcConName]['password']+"@"+config[srcConName]['host']+"/?service_name="+ config[srcConName]['service_name']
connection=connection.replace('"','')
print(connection)
```

```python
date_var=datetime.date.today()

Allfiles=[f for f in listdir(Spl_path) if isfile(join (Spl_path,f))]
for file in Allfiles:
    x=file.split(".")
    print(file)
    fileName=x[0]+"_"+str(date_var)
    print(fileName)
    filepath=r"C:/Users/nisha/OneDrive/Documents/Python_Scripts/Oracle_Data_Extraction/Output/"+fileName+".csv"
    Spl_file_path=r"C:/Users/nisha/OneDrive/Documents/Python_Scripts/Oracle_Data_Extraction/SQLs/"+file
    with open(Spl_file_path,'r',encoding="utf8",errors='ignore') as myfile:
        query=myfile.read()
    engine = sqlalchemy.create_engine(connection)
    df_result = pd.read_sql(query, engine)
    engine.dispose()
    df_result.to_csv(filepath,encoding="utf8",index=False)
    print("done")
    end_time=datetime.datetime.now()
    print(end_time)
    diff_time=end_time-start_time
    print(diff_time)    
```
